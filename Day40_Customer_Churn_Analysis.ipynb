```python
df.groupby("Contract")["Churn"].value_counts()
```

```python
contract_analysis = df.groupby("Contract")["Churn"].value_counts().unstack(fill_value=0)

contract_analysis["Total Customers"] = (
    contract_analysis["Yes"] + contract_analysis["No"]
)

contract_analysis["Churn Rate (%)"] = (
    contract_analysis["Yes"] /
    contract_analysis["Total Customers"] * 100
).round(2)

contract_analysis
```

```python
df["Tenure Group"] = pd.cut(
    df["Tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12 Months", "13-24 Months", "25-48 Months", "49-72 Months"],
    include_lowest=True
)

df[["Tenure", "Tenure Group"]].head(10)
```

```python
tenure_analysis = df.groupby(
    "Tenure Group",
    observed=False
)["Churn"].value_counts().unstack(fill_value=0)

tenure_analysis["Total Customers"] = (
    tenure_analysis["Yes"] + tenure_analysis["No"]
)

tenure_analysis["Churn Rate (%)"] = (
    tenure_analysis["Yes"] /
    tenure_analysis["Total Customers"] * 100
).round(2)

tenure_analysis
```

```python
monthly_charges_analysis = df.groupby("Churn")["MonthlyCharges"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)

monthly_charges_analysis
```

```python
internet_analysis = df.groupby(
    "InternetService"
)["Churn"].value_counts().unstack(fill_value=0)

internet_analysis["Total Customers"] = (
    internet_analysis["Yes"] + internet_analysis["No"]
)

internet_analysis["Churn Rate (%)"] = (
    internet_analysis["Yes"] /
    internet_analysis["Total Customers"] * 100
).round(2)

internet_analysis
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=internet_analysis.index,
    y=internet_analysis["Churn Rate (%)"]
)

plt.title("Customer Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")

plt.show()
```

```python
payment_analysis = df.groupby(
    "PaymentMethod"
)["Churn"].value_counts().unstack(fill_value=0)

payment_analysis["Total Customers"] = (
    payment_analysis["Yes"] + payment_analysis["No"]
)

payment_analysis["Churn Rate (%)"] = (
    payment_analysis["Yes"] /
    payment_analysis["Total Customers"] * 100
).round(2)

payment_analysis
```

```python
plt.figure(figsize=(9, 5))

payment_analysis["Churn Rate (%)"].drop(
    "unknown", errors="ignore"
).plot(kind="bar")

plt.title("Customer Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
monthly_charges_analysis = df.groupby("Churn")["MonthlyCharges"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)

monthly_charges_analysis
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Churn",
    y="MonthlyCharges",
    data=df
)

plt.title("Monthly Charges Comparison by Churn")
plt.xlabel("Customer Churn")
plt.ylabel("Monthly Charges")

plt.show()
```
